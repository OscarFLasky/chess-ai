```python
import time

from chessai.arena import play_match
from chessai.engine.mcts import MCTSEngine
from chessai.engine.policy import PolicyEngine
import chess

CKPT = "../parameters/ckpt_32000.pt"
NB_MATCHS = 20
NB_SIMS = 200
BATCH_SIZE = 32


policy = PolicyEngine(ckpt_path=CKPT)
mcts = MCTSEngine(policy, nbSims=NB_SIMS, batch_size=BATCH_SIZE)

print(
    f"device={policy.device}  sims={NB_SIMS}  batch={BATCH_SIZE}  matchs={NB_MATCHS}",
    flush=True,
)

t0 = time.perf_counter()
res = play_match(policy, mcts, nbmatchs=NB_MATCHS, progress_every=10)

print()
print(f"{res}   ({time.perf_counter() - t0:.0f}s)")
```
